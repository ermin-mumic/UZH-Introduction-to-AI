# Decision Networks — Umbrella Example (Interactive MEU and VPI)
This notebook lets you explore how decisions change with utilities, priors, and forecast accuracy.
All parameters can be modified using sliders; initial values match the lecture slides.

```python
from ipywidgets import interact, FloatSlider
```

## 1. MEU without information
Computes expected utility of leaving or taking based only on prior probability of rain.

```python
def meu_no_info(p_rain,u_ls,u_lr,u_ts,u_tr, verbose=False):
    eu_leave = p_rain*u_lr + (1-p_rain)*u_ls
    eu_take  = p_rain*u_tr + (1-p_rain)*u_ts
    if verbose:
        print(f"EU(leave) = {eu_leave}")
        print(f"EU(take) = {eu_take}")
    return eu_leave, eu_take
```

```python
@interact(p_rain=FloatSlider(min=0,max=1,step=0.01,value=0.3),
          u_ls=FloatSlider(min=-100,max=150,step=1,value=100),
          u_lr=FloatSlider(min=-100,max=150,step=1,value=0),
          u_ts=FloatSlider(min=-100,max=150,step=1,value=20),
          u_tr=FloatSlider(min=-100,max=150,step=1,value=70))

def run_no_info(p_rain,u_ls,u_lr,u_ts,u_tr):
    eu_leave, eu_take = meu_no_info(p_rain,u_ls,u_lr,u_ts,u_tr, verbose=True)
    print('MEU(no info) =', max(eu_leave, eu_take))
```

## 2. MEU given forecast (bad or good)
Uses P(rain|forecast) chosen via sliders.

```python
def meu_posterior(p_rain_post,u_ls,u_lr,u_ts,u_tr, forecast, verbose=False):
    eu_leave = p_rain_post*u_lr + (1-p_rain_post)*u_ls
    eu_take  = p_rain_post*u_tr + (1-p_rain_post)*u_ts
    if verbose:
        print(f"EU(leave | {forecast}) = {eu_leave}")
        print(f"EU(take | {forecast}) = {eu_take}")
    return eu_leave, eu_take
```

```python
@interact(p_rain=FloatSlider(min=0,max=1,step=0.01,value=0.3),
          p_rain_bad=FloatSlider(min=0,max=1,step=0.01,value=0.66),
          p_rain_good=FloatSlider(min=0,max=1,step=0.01,value=0.05),
          p_bad=FloatSlider(min=0,max=1,step=0.01,value=0.41),
          p_good=FloatSlider(min=0,max=1,step=0.01,value=0.59),
          u_ls=FloatSlider(min=-100,max=150,step=1,value=100),
          u_lr=FloatSlider(min=-100,max=150,step=1,value=0),
          u_ts=FloatSlider(min=-100,max=150,step=1,value=20),
          u_tr=FloatSlider(min=-100,max=150,step=1,value=70))

def run_info(p_rain_bad,p_rain_good,u_ls,u_lr,u_ts,u_tr):
    eu_leave_bad, eu_take_bad = meu_posterior(p_rain_bad,u_ls,u_lr,u_ts,u_tr, "bad", verbose=True)
    eu_leave_good, eu_take_good = meu_posterior(p_rain_good,u_ls,u_lr,u_ts,u_tr, "good", verbose=True)
    print('MEU(F=bad) =', max(eu_leave_bad, eu_take_bad))
    print('MEU(F=good) =', max(eu_leave_good, eu_take_good))
```

## 3. Value of Perfect Information (VPI)
Computes expected gain from learning the forecast.

```python
def vpi(p_rain,p_rain_bad,p_rain_good,p_bad,p_good,u_ls,u_lr,u_ts,u_tr):
    eu_l0, eu_t0 = meu_no_info(p_rain,u_ls,u_lr,u_ts,u_tr)
    base = max(eu_l0, eu_t0)
    eu_lb, eu_tb = meu_posterior(p_rain_bad,u_ls,u_lr,u_ts,u_tr, "bad")
    eu_lg, eu_tg = meu_posterior(p_rain_good,u_ls,u_lr,u_ts,u_tr, "good")
    eu_with = p_bad*max(eu_lb,eu_tb) + p_good*max(eu_lg,eu_tg)
    return eu_with - base
```

```python
@interact(p_rain=FloatSlider(min=0,max=1,step=0.01,value=0.3),
          p_rain_bad=FloatSlider(min=0,max=1,step=0.01,value=0.66),
          p_rain_good=FloatSlider(min=0,max=1,step=0.01,value=0.05),
          p_bad=FloatSlider(min=0,max=1,step=0.01,value=0.41),
          p_good=FloatSlider(min=0,max=1,step=0.01,value=0.59),
          u_ls=FloatSlider(min=-100,max=150,step=1,value=100),
          u_lr=FloatSlider(min=-100,max=150,step=1,value=0),
          u_ts=FloatSlider(min=-100,max=150,step=1,value=20),
          u_tr=FloatSlider(min=-100,max=150,step=1,value=70))

def run_vpi(p_rain,p_rain_bad,p_rain_good,p_bad,p_good,u_ls,u_lr,u_ts,u_tr):
    eu_leave, eu_take = meu_no_info(p_rain,u_ls,u_lr,u_ts,u_tr)
    print('MEU(no info) =', max(eu_leave, eu_take))

    eu_leave_bad, eu_take_bad = meu_posterior(p_rain_bad,u_ls,u_lr,u_ts,u_tr, "bad")
    eu_leave_good, eu_take_good = meu_posterior(p_rain_good,u_ls,u_lr,u_ts,u_tr, "good")
    print('MEU(F=bad) =', max(eu_leave_bad, eu_take_bad))
    print('MEU(F=good) =', max(eu_leave_good, eu_take_good))

    print('VPI:', vpi(p_rain,p_rain_bad,p_rain_good,p_bad,p_good,u_ls,u_lr,u_ts,u_tr))
```
